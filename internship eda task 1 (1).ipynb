#### Find numbers with an Even number of Digits

```python
n=list(map(int,input().split()))
c=0
for i in range(len(n)):
    if len(str(n[i]))%2==0:
        c+=1
print(c)
```

#### Number of Good Pairs

```python
n=list(map(int,input().split()))
c=0
for i in range(len(n)):
    for j in range(1+i,len(n)):
        if n[i]==n[j]:
            c+=1
print(c)
```

#### How Many Number Are Smaller Than the Current Number

```python
nums=list(map(int,input().split()))
r=[]
for i in range(len(nums)):
    c=0
    for j in range(len(nums)):
        if nums[i]>nums[j]:
            c+=1
    r.append(c)
r
```

#### Subtract the Product and Sum of Digits of an Integer

```python
n=int(input())
p,s=1,0
while n>0:
    p*=n%10
    s+=n%10
    n=n//10
p-s
```

#### XOR Operation in an Array

```python
n=int(input())
start=int(input())
nums=[]
for i in range(n):
    a=start+2*i
    nums.append(a)
b=0
for i in nums:
    b^=i
b
```
